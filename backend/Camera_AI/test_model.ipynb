```python
import cv2 
from ultralytics import YOLO
```

```python
video_path = r'D:\Share_Projects\AutomatedChickenFarmManagement\backend\Camera_AI\Data\Test_data\Videos\video_test.mp4'
model_path = r'D:\Share_Projects\AutomatedChickenFarmManagement\backend\Camera_AI\runs\detect\runs\my_project\yolo_model-2\weights\best.pt'

model = YOLO(model_path)  # model đã train
cap = cv2.VideoCapture(video_path)
ret = True 
while ret: 
    ret,frame = cap.read()
    if not ret:
        break

    #chạy model yolo 
    results = model(frame, conf= 0.75)
    # Hiển thị kết quả (YOLO có sẵn hàm plot)
    annotated_frame = results[0].plot()

    cv2.imshow("YOLO Detection", annotated_frame)

    if cv2.waitKey(1) & 0xFF == ord("q"):
        break
cap.release()
cv2.destroyAllWindows()
```
